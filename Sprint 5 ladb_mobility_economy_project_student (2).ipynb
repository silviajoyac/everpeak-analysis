### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

````python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc'] = pd.to_datetime(traffic["UpdateTimeUTC"], errors="coerce")
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic["UpdateTimeUTCWeekAgo"], errors="coerce")

# verificar el cambio
traffic.info()
````

````python
# Limpia separadores y convierte columnas numéricas en eco

eco["city_gdp_capita"] = (eco["City GDP/capita"]
    .str.replace(".", "", regex=False)
    .str.replace(",", ".", regex=False)
    .astype("float"))






eco['unemployment_pct'] = (eco['Unemployment %']
                          .str.replace('%', '')
                          .str.replace(',', '.')
                          .astype(float))
eco['population_m'] = (eco['Population (M)']
                         .str.replace(',', '.')
                          .astype(float))

# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco['population_m'] * 1_000_000

# verificar el cambio
eco.info()
eco.head(3)
````

### 3.1 Extraer columna año y filtrar 2024

**🎯Objetivo**
Identificar el año de cada registro y mantener solo los registros del 2024.

**Intrucciones**

- Como el DataFrame `traffic` no tiene una columna de año, utiliza el atributo `.dt.year` sobre su columna de fecha para crear una nueva columna llamada `year`.
- Filtra las filas donde el año sea **2024**.
- Utiliza `.copy()` para crear dos nuevos DataFrames (`traffic_2024` y `eco_2024`) para evitar modificar el dataset original.

````python

# Filtra los registros del año 2024
traffic_2024 = traffic[traffic['year'] == 2024].copy()
eco_2024 = eco[eco['Year'] == 2024].copy()

# Revisar dataframes nuevos
display(traffic_2024.head())
display(eco_2024.head())
````

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

````python


# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 =  (traffic_2024
                          .groupby(['City', 'Country', 'year'], as_index=False)
                          .agg({
                              'JamsDelay': 'mean',
                              'TrafficIndexLive': 'mean',
                              'JamsLengthInKms': 'mean',
                              'JamsCount': 'mean',
                              'MinsDelay': 'mean',
                              'TravelTimeLivePer10KmsMins': 'mean',
                              'TravelTimeHistoricPer10KmsMins': 'mean' }))


# Mostrar resultado
traffic_city_year_2024.head()
````

## 🧩Paso 5: Unir movilidad y economía

Combinar datasets te permite analizar cómo se relacionan los indicadores económicos con los de movilidad.

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.
- Mantén solo las ciudades y años presentes en ambos datasets.
- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

````python
# Seleccionar columnas clave de tráfico y economía
left_cols = ['City','Country','year','JamsDelay','TrafficIndexLive',
             'JamsLengthInKms','JamsCount','MinsDelay',
             'TravelTimeLivePer10KmsMins','TravelTimeHistoricPer10KmsMins']

right_cols = ['City','Year','city_gdp_capita','unemployment_pct','PM2.5 (μg/m³)','population']

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()



eco_2024_small = eco_2024_small.rename(columns={
    'Year': 'year'})


# Unir datasets
merged = traffic_2024_small.merge(eco_2024_small,
                                  on=['City', 'year'],
                                  how='inner')

# Mostrar las primeras 5 filas
print(merged.head(5))
````

**Contexto & objetivo:**  
- Responde la pregunta central del análisis: ¿qué relación existe entre la movilidad urbana (congestión, tiempos de viaje) y la productividad económica (PIB per cápita)?
- Explica brevemente las variables clave utilizadas y su relevancia para la toma de decisiones.

- Este análisis examina 34,334 registros de tráfico diario con indicadores económicos de 15 ciudades latinoamericanas en 2024. Las variables clave utilizadas son: JamsDelay (retrasos en minutos por congestión, proxy de eficiencia de movilidad), TrafficIndexLive (índice de congestión en tiempo real), city_gdp_capita (PIB per cápita en USD, indicador de productividad), unemployment_pct (tasa de desempleo) y PM2.5 (calidad del aire). Estos indicadores son relevantes para decisiones de inversión porque permiten identificar ciudades donde mejorar transporte urbano generaría mayores retornos económicos y reducción de costos sociales por congestión.

**Cobertura de datos:**  
- Especifica los años analizados, número de ciudades y países incluidos.

- El análisis cubre 2024 completo, incluyendo 15 ciudades, 7 países: Brasil (9 ciudades: São Paulo, Rio de Janeiro, Curitiba, Recife, Belo Horizonte, Salvador, Porto Alegre, Brasilia, Fortaleza), Colombia (Bogotá), Perú (Lima), Chile (Santiago), Argentina (Buenos Aires), México (Ciudad de México) y Uruguay (Montevideo). Con estos datos podemos ver que Brasil concentra el 60% de los registros (20,587), permitiendo análisis regional confiable, mientras que los otros países tienen aproximadamente 2,287 registros cada uno. La distribución de datos es suficiente para detectar patrones a nivel de ciudad, aunque genera limitaciones para generalizaciones nacionales para otros países no brasileños.

**Metodología (alto nivel):**  
- Describe los procesos principales: limpieza de datos (formatos, estandarización de columnas).
- Explica la agregación por ciudad–año y el uso de una unión INNER para integrar tráfico y economía.
- Menciona las validaciones visuales empleadas (distribuciones, outliers, tendencias generales).
  
- Para la limpieza de datos: Se estandarizaron tipos de datos de ambas fuentes, convirtiendo a float columnas numéricas con formatos mixtos (PM2.5 con comas decimales, PIB con separadores). Se eliminaron caracteres especiales y se normalizaron nombres de ciudades a minúsculas para consistencia. Agregación: Se utilizó groupby por ciudad-año para consolidar miles de registros diarios de tráfico en promedios comparables con indicadores económicos anuales, reduciendo la dimensión de 34,334 a 15 filas (una por ciudad). Integración: Se aplicó merge INNER en las claves city y year, priorizando registros presentes en ambas bases de datos (tráfico y economía). Validaciones visuales: Se emplearon boxplots para detectar distribuciones sesgadas y outliers extremos en retrasos de tráfico (máximos >10,000 minutos), histogramas para analizar distribuciones de PIB, y cálculo de correlaciones de Pearson entre variables clave. Se identificaron outliers con métodos de rango intercuartílico para revisar anomalías.

**Hallazgos iniciales:**  
- Resume los patrones más importantes entre índices de tráfico y PIB per cápita.
- Destaca anomalías u outliers que podrían requerir revisión adicional o un análisis más profundo.

- El patron principal: La correlación agregada entre JamsDelay y PIB per cápita es débil (0.1624), indicando que la congestión NO es predictor del desarrollo económico. Sin embargo, el análisis por ciudad revela patrones más complejos:

México City (PRIORIDAD EXTREMA): Lidera con 2,833 minutos promedio de retrasos (47 horas/mes), PIB de $21,111 per cápita. Pese a ingresos moderados, la congestión es catastrófica, sugiriendo que la inversión en transporte destrabará potencial económico significativo. Población: 22.1M.
Santiago (PRIORIDAD ALTA): Segunda en retrasos (630 minutos), pero con PIB muy bajo de $2,277 (probable anomalía en datos), población 7.1M. Requiere validación urgente pero potencial de mejora es evidente.
Bogotá (PRIORIDAD ALTA): 1,142 minutos promedio de retrasos, PIB de $11,442, desempleo de 10 porciento con una población 11.3M. Sufre congestión severa comparable a ciudades más ricas, indicando ineficiencia estructural.
Lima (PRIORIDAD MEDIA-ALTA): 1,052 minutos promedio, PIB $13,472 (mejor que Bogotá), desempleo 6.5%, población 11.2M. Congestión severa pero mejores indicadores económicos.
Buenos Aires (MENOR PRIORIDAD): 571 minutos de retrasos, PIB alto de $18,117, desempleo 7.2%, población 15.4M. Mejor manejo de congestión pese a tamaño grande; ciudadanos están acostumbrados a tiempos de viaje más largos.

- Las anomalías destacadas: Algunos registros de JamsDelay superan 10,000 minutos (máximo 15,503), equivalente a 10+ días de congestión continua en registros únicos. Probable causa: errores de captura o eventos especiales (cierres viales). Valores de PIB en Santiago ($2,277) parecen inconsistentes comparado con otras ciudades de similar desarrollo; requiere validación de fuente.

**Recomendaciones**  
Aterriza los hallazgos en acciones: ciudades prioritarias, necesidad de validar fuentes, requerimiento de análisis adicionales, o propuestas de inversión.

- ¿Qué ciudad : Bogotá, Lima o Buenos Aires o alguna otra en particular, muestra la mayor correlación significativa entre altos niveles de congestión vehicular y bajos indicadores de productividad económica, sugiriendo ser una ciudad prioritaria para inversión en infraestructura de transporte?

- Bogotá emerge como ciudad prioritaria sobre Lima y Buenos Aires por la siguiente razón: sufre congestión severa (1,142 min/mes) comparable a cidadesmás desarrolladas, pero con PIB per cápita moderado-bajo ($11,442) y desempleo alto (10%)

Esto indica que la congestión es una barrera estructural al desarrollo económico, no consecuencia de él. Inversión en transporte masivo (metro, BRT) generaría retornos máximos: (a) descongestión beneficiaría 11.3M personas, (b) reduciría tiempos de viaje, aumentando productividad laboral, (c) atraería inversión privada en zonas próximas a transporte. Presupuesto estimado: $5-8B USD para sistema de transporte masivo de 100+ km de líneas.

- Acciones inmediatas: México City: Urgente auditoría de datos y validación de mediciones de tráfico. Si los 2,833 minutos/mes son correctos, requiere intervención federal inmediata: expansión de metro (actualmente insuficiente), carriles BRT dedicados, restricción vehicular por matrícula.
Santiago y Lima: Validar datos de PIB (Santiago muestra valor anómalo). Si confirmados, priorizar ciclovías, carriles preferenciales de buses y semáforos inteligentes antes de inversión en metro (menor costo, impacto rápido).
Buenos Aires: Mantener monitoreo; ciudadanos adaptos a tiempos largos. Enfocarse en optimización de rutas de transporte público en lugar de expansión.

- En conclusión: Existe relación débil entre tráfico y PIB agregado, pero patrones por ciudad revelan que congestión severa con bajo PIB (Bogotá, México City) indica oportunidad de inversión de alto impacto. Bogotá es la ciudad prioritaria porque invertir en transporte urbano sería catalizador del desarrollo económico, no consecuencia de él.
